## Import modules

```python
import os
import sys
from pathlib import Path

cwd = Path.cwd().resolve()
repo_root = next((p for p in [cwd, *cwd.parents] if p.name == 'pips-design-toolkit'), None)
if repo_root is None:
    raise RuntimeError('Run this notebook from within the pips-design-toolkit repository.')
tools_dir = repo_root / "tools"
os.chdir(tools_dir)
if str(tools_dir) not in sys.path:
    sys.path.insert(0, str(tools_dir))
from utils.variables import data_folder
from get_pLM_zeroshot_scores_embeddings import PLMZero
```

## Specify input parameters
* *pLM model type* (`model_type`): e.g. 'esm2'
* *Data subfolder name* (`data_subfolder`): Subfolder name for saving data to. Can be an empty string.
* *Input fasta or CSV filename* (`input_file`): Fasta or CSV filepath in data folder for obtaining sequences for pLM processing.
* *MSA filepath* (`msa_fpath`): Multiple Sequence Alignment filepath for found in 'data/msa/' directory. Reference sequence should be the first entry in the MSA. Only applicable to pLMs that require an MSA input (e.g. PoET). 
* *Bool indicating whether to calculate scores for mutants of a base sequence* (`get_mutations`): True or False. Default=True
* *Dict indicating what embeddings to obtain* (`get_embeddings`):  E.g. {'res_all': True, 'res_mut': True,'res_avg': True}
* *Bool indicating whether sequence list is to be deduplicated before pLM processing* (`deduplicate_sequences`): True or False
* *Bool indicating whether to skip pLM processing (if processed file already exist for subsequent parsing)* (`skip_plm_processing`): True or False. Default=False
* *Bool indicating whether to skip obtaining of logit features (i.e. if we are only obtaining embeddings, or processing wildtype sequences and not mutants)* (`skip_logits_features`): True or False. Default=False
* *Batch size of sequences fed to pLM for processing* (`batch_size`): Default=1
* *Name of CSV column in 'input_file' listing base sequence name* (`csv_name_col`): Default='protein_name'
* *Separator string for joining mutations in combinatorial mutant* (`sep`): Default='+', e.g. 'A21V+E50K'

```python
model_type = 'esm2'
data_subfolder = ''
input_file = 'GOh1052.fasta'
msa_fpath = None
get_mutations=True
get_embeddings =  {'res_all': True, 'res_mut': True,'res_avg': True} # {'res_all': False, 'res_mut': False,'res_avg': False} # 
deduplicate_sequences = True
skip_plm_processing=False
skip_logits_features=False
batch_size = 1
csv_name_col = 'protein_name'
sep = '+'
```

## Get zero-shot scores with PLM 

Log-likelihood scores are saved at: './data_folder/conservation_and_distance/<`seq_name`>_<`model_name`>_LLRsum_entropy.csv'

PLM embeddings are saved at: 
* './data_folder/feature_extraction/<`seq_name`>_<`model_name`>_mut_embeddings.csv'
* './data_folder/feature_extraction/<`seq_name`>_<`model_name`>_seq_embeddings.csv'

```python
# initialize PLMZero class
plm_zeroshot = PLMZero(
    model_type,
    msa_fpath,
    data_folder,
    data_subfolder,
    get_mutations,
    get_embeddings,
    deduplicate_sequences,
    skip_plm_processing,
    skip_logits_features
    )

# get PLM inputs
inputs_all = plm_zeroshot.get_inputs(input_file, batch_size, csv_name_col, sep)

# process through pLM
plm_zeroshot.run_pLM_processing(inputs_all)

# parse results for each sequence in inputs_all
plm_zeroshot.parse_plm_results(inputs_all, get_llr_entropy_plot=False)
```
