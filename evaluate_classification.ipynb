## Specify input parameters
* *pLM model type* (`model_type`): e.g. 'esm2'
* *Data subfolder name* (`data_subfolder`): Subfolder name for saving data to. Can be an empty string.
* *Input CSV filename* (`input_file`): CSV filepath in data folder for obtaining sequences for pLM processing.
* *MSA filepath* (`msa_fpath`): Multiple Sequence Alignment filepath for found in 'data/msa/' directory. Reference sequence should be the first entry in the MSA. Only applicable to pLMs that require an MSA input (e.g. PoET). 
* *Bool indicating whether to calculate scores for mutants of a base sequence* (`get_mutations`): True or False
* *Dict indicating what embeddings to obtain* (`get_embeddings`):  E.g. {'res_all': True, 'res_mut': True,'res_avg': True}
* *Bool indicating whether sequence list is to be deduplicated before pLM processing* (`deduplicate_sequences`): True or False
* *Bool indicating whether to skip pLM processing (if processed file already exist for subsequent parsing)* (`skip_plm_processing`): True or False. Default=False
* *Bool indicating whether to skip obtaining of logit features (i.e. if we are only obtaining embeddings, or processing wildtype sequences and not mutants)* (`skip_logits_features`): True or False. Default=False
* *Batch size of sequences fed to pLM for processing* (`batch_size`): Default=1
* *Name of CSV column in 'input_file' listing base sequence name* (`csv_name_col`): Default='protein_name'
* *Separator string for joining mutations in combinatorial mutant* (`sep`): Default='+', e.g. 'A21V+E50K'
