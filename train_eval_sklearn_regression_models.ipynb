## Specify input parameters
* *Data subfolder name* (`data_subfolder`): Subfolder name for saving data to. Can be an empty string.
* *Input CSV directory* (`input_csv_dir`): Subfolder containing input CSV data (mutations and labels to train on)
* *Input CSV filename* (`input_file`): CSV filepath in data folder for obtaining sequences for pLM processing.
* *Y labels to train on* (`ylabel_list`): List of labels for training regression model on
* *Type of splits to evaluate* (`splits_to_evaluate`): List of split types, e.g. 'random', 'mutres-modulo', 'protein', 'custom'. Default=['random']
* *List of extra columns to include in input dataframe, besides X and Y features* (`extra_cols_to_get`): Default=['protein_name', 'mutations']
* *Output filename of CSV storing regression results* (`output_fname`): str
* *Prefix of filename for storing intermediate files (e.g. full aggregated data)* (`fname_prefix`): str
* *List of suffixes of input data files to obtain; only applicable if mutations are reversed* (`data_suffix_list`): Default=['']; with mutation reversals=['','_rev']
